***fetching data from api***

api= data pipelines. transfer of information

```python
import pandas as pd
import requests
```

```python
response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=587f5fd0aa118783cb93d12e2fb5d84b&language=en-US&page=1')
```

```python
response.json()
```

```python
response.json()['results']
```

```python
pd.DataFrame(response.json()['results'])
```

```python
pd.DataFrame(response.json()['results'],columns=['id','title','overview','release_date','popularity'])
```

```python
df=pd.DataFrame()
```

```python
for i in range(1,501):
  response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=587f5fd0aa118783cb93d12e2fb5d84b&language=en-US&page={}'.format(i))

  # Check if the request was successful
  if response.status_code == 200:
    json_data = response.json()
    if 'results' in json_data:
      temp_df=pd.DataFrame(json_data['results'],columns=['id','title','overview','release_date'])
      df=pd.concat([df, temp_df], ignore_index=True)
    else:
      print(f"Warning: 'results' key not found in response for page {i}. Full response: {json_data}")
  else:
    print(f"Error: API request failed for page {i} with status code {response.status_code}. Response: {response.text}")
```

```python
df.shape
```

```python
df.to_csv('movies.csv')
```
