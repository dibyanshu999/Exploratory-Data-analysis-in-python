***Bivariate analysis*** : analysis of 2 variables

````python
import seaborn as sns
import pandas as pd
````

````python
tips=sns.load_dataset('tips')  #no need to upload file
````

````python
tips
````

````python
titanic=sns.load_dataset('titanic')  #no need to upload file
````

````python
iris=sns.load_dataset('iris')
````

````python
flights=sns.load_dataset('flights')
````

1. scatterplot  (numerical-numerical)

````python
sns.scatterplot(x=tips['total_bill'], y=tips['tip'])
````

````python
sns.scatterplot(x=tips['total_bill'], y=tips['tip'],hue=tips['sex'],style=tips['smoker'])
````

2.barplot

````python
titanic.head()
````

````python
sns.barplot(x=titanic['pclass'],y=titanic['age'],hue=titanic['sex'])
````

3.boxplot (numerical-categorical)

````python
sns.boxplot(x=titanic['sex'],y=titanic['age'])
````

4.distplot(numerical-categorical)

````python
sns.distplot(titanic[titanic['survived']==0]['age'])
````

5.heatmap(categorical-categorical)

````python
sns.heatmap(pd.crosstab(titanic['pclass'],titanic['survived']))
````

````python
d=(titanic.groupby('pclass')['survived'].mean()*100)
````

````python
d.plot(kind='bar')
````

6.clustermap(categorical-categorical)

````python
sns.clustermap(pd.crosstab(titanic['sibsp'],titanic['survived']))
````

8.lineplot( numerical-numerical)

````python
new=flights.groupby('year')['passengers'].sum().reset_index()
````

````python
sns.lineplot(x=new['year'],y=new['passengers'])
````

````python
sns.clustermap(flights.pivot_table(values='passengers',index='month',columns='year'))
````
